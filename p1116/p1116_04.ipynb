랜덤포레스트로 붓꽃데이터

```python
df=pd.read_csv('data/iris.csv')
df.head(2)
```

```python
# data=df.iloc[:,1:-1].values
# label=df.iloc[:,-1].values

# data,label 저장
data = df[df.columns[1:-1]].values
label = df['Species'].values
```

```python
print(data.shape)   # 예상: (150, 4)
print(label.shape)  # 예상: (150,)
```

```python
# 데이터 분리
train_data, test_data, train_label, test_label=train_test_split(data, label, stratify=label, random_state=42 )
```

```python
# 정규화 작업을 안해도 됨. / 기본이 100번
rf= RandomForestClassifier(n_jobs=-1, n_estimators=100)
rf.fit(train_data,train_label)

print('train정확도:', rf.score(train_data,train_label))
print('test정확도:', rf.score(test_data,test_label))

```

```python
rf=RandomForestClassifier(n_estimators=100)
scores = cross_validate(rf,train_data, train_label, return_train_score=True, n_jobs=-1)
print('train정확도:', np.mean(scores['train_score']))
print('test정확도:', np.mean(scores['test_score']))
```
