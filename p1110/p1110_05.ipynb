```python
# 그래프 관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

# 판다스 관련 import
import pandas as pd
import numpy as np

# 인공지능 import
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier  #분류
from sklearn.neighbors import KNeighborsRegressor   #예측
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error # 평균절대값오차 오차범위
from sklearn.preprocessing import StandardScaler # 정규화 클래스
from sklearn.model_selection import train_test_split
```

```python
# 1. 공부시간에 따른 시험정수를 학습해서 시간을 입력하면 점수를 예측하시오
# scatter그래프에 점수와 직선을 표시하시오
```

```python
df=pd.read_csv('data/study.csv', sep='\t')
df.head()
```

```python
#  vlaues를 안하면 Series(1차원 데이터, 문자열), 붙이면 np.array(1차원 데이터 , 숫자)

# data=df['our'].values
# label=df['score'].values

# data=df.iloc[:,0].values
# label=df.iloc[:,1:].values


data=df.iloc[:,:-1].values
label=df.iloc[:,-1:].values
```

```python
# hour=np.array([0.5,1.2,1.8,2.4,2.6,3.2,3.9,4.4,4.5,5,5.3,5.8,6,6.1,6.2,6.9,7.2,8.4,8.6,10])
# score=np.array([10,8,14,26,22,30,42,48,38,58,60,72,62,68,72,58,76,86,90,100])
```

```python
train_data, test_data, train_label, test_label= train_test_split(data, label)
```

```python
# # 1차원 데이터를 2차원 데이터로 만들기
# train_data= train_data.reshape(-1,1)
# test_data= test_data.reshape(-1,1)
```

```python
# 예측 모델로 학습
lr=LinearRegression()
lr.fit(train_data, train_label)

# 예측
predict= lr.predict(test_data)
print('예측점수:',predict)

# 정확도
score=lr.score(train_data,train_label)
score2=lr.score(test_data,test_label)
print('정확도:',score)
print('정확도:',score2)

# 기울기
print('기울기:', lr.coef_)
print('y절편:', lr.intercept_)
```

```python
in_data= input('공부한 시간을 입력하시오(예:7.9,4.2)')
in_data=float(in_data)
predict= lr.predict([[in_data]])
print('성적:',predict)


plt.scatter(data,label)
plt.scatter(in_data, predict,marker='^')
plt.plot([0.5,10],[0.5*lr.coef_[0]+lr.intercept_,10*lr.coef_[0]+lr.intercept_])

plt.xlabel('시간')
plt.ylabel('성적')
plt.show()
```
