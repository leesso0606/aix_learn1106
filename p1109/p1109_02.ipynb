```python
from sklearn import model_selection,svm, metrics
import pandas as pd
import numpy as np
from PIL import Image,ImageOps
```

```python
#  데이터 가져오기
# header=None: 첫 줄부터 데이터이며, 열 이름을 적은 줄은 없다
train_csv= pd.read_csv('data/train.csv', header=None)
test_csv=pd.read_csv('data/t10k.csv', header=None)
```

```python
# 픽셀의 정규화/ 0~225 숫자를 0-1 사이의 숫자로 정규화
# 이미지 한 장의 픽셀값->28 × 28 이미지 → 픽셀값 784개 → CSV의 한 행에 저장
def test(x):
    output = []
    for i in x:
        output.append(float(i)/256)
    return output
```

```python
#  train_csv[0] : 행을 뽑아옴

# 데이터 정규화:  0-255/
# 열을 뽑아옴(0번째 제외 1번째부터 가져와라)
# train_csv.iloc[행선택, 열선택]: 모든행, 1번째 열부터 가져와라/ values: 값을 배열로 가져와라
# 가져온 값을 위 test()함수에 적용하여 list로 변환
# map(적용할 함수, 값들의 모음): 여러 값에 같은 함수를 하나씩 적용하는 도구
train_csv_data=list(map(test,train_csv.iloc[:,1:].values))
test_csv_data=list(map(test,test_csv.iloc[:,1:].values))
train_csv_data[0]
```

```python
#  정답 label작업
train_csv_label = train_csv[0].values
train_csv_label
test_csv_label = test_csv[0].values
test_csv_label
```

```python
# 데이터 학습 : 다른 알고리즘 사용해도 상관없음.
clf=svm.SVC()
clf.fit(train_csv_data, train_csv_label)
# 데이터 예측
result= clf.predict(test_csv_data)
# 정답율
score= metrics.accuracy_score(test_csv_label, result)
print('정답율:',score)
```

```python
# 이미지 열기, .convert("L"):흑백으로 변환
img=Image.open('data/eight.png').convert("L")
# 색상반전
img = ImageOps.invert(img)
```

```python
# 숫자가 있는 영역만 추출
bbox= img.getbbox()
img = img.crop(bbox)

# 비율축소 20*20
img.thumbnail((20,20),Image.Resampling.LANCZOS)

# 배경사이즈 28*28
new_img = Image.new('L',(28,28),0)

# 배경사이즈에 숫자를 가운데 붙여넣기
x=(28-img.width)//2
y=(28-img.height)//2
new_img.paste(img,(x,y))

```

```python
# 예측용 데이터-0-225
# 이미지를 실수 형태의 숫자 배열로
img_data = np.array(new_img,dtype=np.float32)

# reshape(1,784): 28×28 배열을 1행 × 784열로 펼침, 현재 데이터처럼
flat_img_data = img_data.reshape(1,784)/256.0
```

```python
# 데이터 예측 :  항상 2차원형태여야함.
result= clf.predict(flat_img_data)
print(result)

```
