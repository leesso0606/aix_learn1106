```python
# 4.2,1,Cafe 일때 예상점수를 구하시오.
```

```python
df=pd.read_csv('data/MultipleLinearRegressionData.csv')
df.head(2)
```

```python
data= df.iloc[:,:2].values
label = df.iloc[:,-1].values
label
```

```python
#  문자열 인코딩하기
place= df['place'].to_numpy().reshape(-1,1)

# 문자 형태의 범주를 숫자 0과 1로 바꾸는 도구를 만드는 코드
# OneHotEncoder: 지역, 색상 같은 범주를 각각 0과 1로 표현
# sparse_output=False: 결과를 일반적인 NumPy 배열 형태로 반환
# drop='first': 범주 중 첫 번째 범주에 해당하는 열을 제외
encoder = OneHotEncoder(sparse_output=False, drop='first')

# fit_transform() : place에 어떤 범주가 있는지 파악하여 파악한 범주를 기준으로 데이터를 0과1로 변환
result = encoder.fit_transform(place)

```

```python
data[:2]
```

```python
result[:5]
```

```python
# data + result 합치기
total_data = np.concatenate((data,result),axis=1)
total_data[:2]

# 같은말임
# total_data2 =np.column_stack((data,result))
# total_data2
```

```python
# 학습데이터,테스트데이터 분리
train_data,test_data,train_label,test_label = \
train_test_split(total_data,label,random_state=42)
train_data[:2]
```

```python
# 모델학습
lr=LinearRegression()
lr.fit(train_data, train_label)

# 정확도
score=lr.score(train_data, train_label)
score2=lr.score(test_data, test_label)
print('train 정확도 : ',score)
print('test 정확도 : ',score2)
```

```python
# 예측
predict = lr.predict(test_data)
print(predict)
```
