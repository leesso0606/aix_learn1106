```python
# 데이터 연결
fish = pd.read_csv('data/fish_logistic.csv')
fish.head(2)
# fish.info()
# fish.describe()
fish.shape
```

```python
# 컬럼 : Species	Weight	Length	Diagonal	Height	Width
# 생선 종류: ['Bream', 'Roach', 'Whitefish', 'Parkki', 'Perch', 'Pike', 'Smelt'], Length: 7, dtype: str

# 데이터 분리
data= fish[fish.columns[1:]].values
lable= fish['Species'].values
data[:2]

# 데이터 분리
train_data, test_data, train_label, test_label = train_test_split(data, lable, random_state=42)
```

```python
# 정규화
ss=StandardScaler()
ss.fit(train_data)

train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)
```

```python
# lr = LogisticRegression()
# sc=  SGDClassifier(): 분류
# loss='log_loss': lr = LogisticRegression()에 있는 log함수를 사용한다.
# max_iter=10 : 10번 반복
# random_state=42 : 랜덤선택을 고정 ( 실제 테스트때는 없어야함, 연습할 때 랜덤을 변경하면 정확도도 함께 변경되어서 고정한것.)
# verbose=1 : 반복할 때 정보를 보여줌. / 설정하지 않으면 보여주지 않음.


sc=  SGDClassifier(loss='log_loss', max_iter=10, random_state=42)
sc.fit(train_scaled, train_label)

print("train정확도:", sc.score(train_scaled, train_label))
print("test정확도:",sc.score(test_scaled, test_label))
```

```python
# 동일한 훈련을 반복
# fit(): 처음부터 시작, 모든 기울기와 절편을 리셋해서 처음부터 학습 시작
# partial_fit(): 현재 있는 정보에서 학습시작(누적)

sc.partial_fit(train_scaled, train_label)

print("train정확도:", sc.score(train_scaled, train_label))
print("test정확도:",sc.score(test_scaled, test_label))
```
