```python
df=pd.read_csv('data/iris.csv')
df.head(2)
```

```python
df['Species'].unique()
```

```python
# 붓꽃품종분류: 89.23% 확률로 Iris-setosa 품종으로 판별됩니다. 로 출력되게 하시오
```

```python
data= df[df.columns[1:-1]].values
label= df['Species'].values

data[:1]
```

```python
# 학습/테스트 데이터 분리
train_data, test_data, train_label, test_label = train_test_split(data, label, stratify=label, random_state=42)
```

```python
train_data[:1]
```

```python
# 정규화
ss=StandardScaler()
ss.fit(train_data)

train_scaled= ss.transform(train_data)
test_scaled= ss.transform(test_data)
```

```python
# LogisicRegression()
# C(대문자표기, 매개변수)- alpah, C 구제, C가 높아질수록 완화, 낮아질수록 강화, 디폴트가 20
# alpah for문으로 그래프로 가장 최적화된 구제를 찾아서 적용.
# max_iter 반복을 자동화해주는 매개변수

lr=LogisticRegression()
lr.fit(train_scaled, train_label)

# 예상
print(lr.predict(test_scaled))

# 정확도
print("train정확도:",lr.score(train_scaled, train_label))
print("test정확도:",lr.score(test_scaled, test_label))
```

```python
print(lr.predict(test_scaled[:1]))
lr.predict_proba(test_scaled[:1])
```

```python
# 붓꽃품종분류: 89.23% 확률로 Iris-setosa 품종으로 판별됩니다. 로 출력되게 하시오
```

```python
in_data='5.1,3.5,1.4,0.2'
in_data=list(map(float,in_data.split(',')))

# 정규화
ss=StandardScaler()
ss.fit(train_data)

in_data_scaled= ss.transform([in_data])

# 모델학습
lr=LogisticRegression()
lr.fit(train_scaled,train_label)

# 예측- 'Iris-setosa', 'Iris-versicolor', 'Iris-virginica'
result=lr.predict(in_data_scaled)
result2 = lr.predict_proba(in_data_scaled)

# 확율

if result[0]=='Iris-setosa':
    result3=result2[0][0]
elif result[0]=='Iris-versicolor':
    result3=result2[0][1]
elif result[0]=='Iris-virginica':
    result3=result2[0][2]


print(f"붓꽃품종분류: {result3*100:.2f}% 확률로 {result[0]} 품종으로 판별됩니다.")
```
