```python
# 그래프 관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

# 판다스 관련 import
import pandas as pd
import numpy as np

# 인공지능 import
from sklearn import svm,metrics # 분류와 숫자 예측에 사용하는 모델 모음,모델의 예측 성능을 평가하는 도구 모음
from sklearn.neighbors import KNeighborsClassifier  #분류(가까운 이웃 데이터를 참고해 종류를 분류)
from sklearn.neighbors import KNeighborsRegressor   #예측(가까운 이웃데이터를 참고해 숫자를 예측)
from sklearn.linear_model import LinearRegression   #예측 (입력과 결과의 선형관례를 학습해 숫자를 예측)
from sklearn.preprocessing import PolynomialFeatures # 다항특성 (x², x₁ × x₂ 같은 특성을 추가)
from sklearn.preprocessing import OneHotEncoder      # 원핫인코딩(문자형 번주를 0과1로 변환)
from sklearn.linear_model import Ridge              #릿지 (계수가 지나치게 커지는 것을 억제하는 선형 회귀)
from sklearn.linear_model import Lasso              #라쏘 (일부 계수를 0으로 만들어 특성 선택에도 활용하는 선형 회귀)

from sklearn.metrics import mean_absolute_error # 평균절대값오차 오차범위 (실제값과 예측값의 차이를 절댓값으로 구한 뒤 평균 계산)
from sklearn.preprocessing import StandardScaler # 표준화 (각 특성을 평균 0, 표준편차 1 기준으로 표준화)
from sklearn.model_selection import train_test_split #(데이터를 학습용과 평가용으로 분리)
```

```python
# LinearRegression(): 예측: 지속적인 데이터를 가지고 확률을 알아내는 것
# 분류 : 1,0/빙어,도비 구분

df=pd.read_csv('data/perch_full.csv')
df.head(2)
```

```python
# 무게 데이터 (label)
perch_weight = [5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0,
    110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0,
    130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0,
    197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0,
    514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0,
    820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0,
    1000.0, 1000.0]
```

```python
# 데이터 전처리
# 학습/테스트 데이터 분리
train_data, test_data, train_label, test_label= train_test_split(df,perch_weight,random_state=42)
```

```python

# 데이터학습
lr=LinearRegression()
lr.fit(train_data, train_label)

# 데이터 예측
predict= lr.predict(test_data)
print('예측무게:',predict)

# 정확도
print('train 정확도:',lr.score(train_data, train_label))
print('test 정확도:',lr.score(test_data, test_label))
```

```python
train_data.shape
```

```python
# 특성을 추가하여 정확도를 높이는 작업
poly=PolynomialFeatures(degree=2, include_bias=False)
poly.fit(train_data)

train_poly=poly.transform(train_data)
test_poly=poly.transform(test_data)

train_poly.shape
```

```python
print(train_data.shape)
print(train_poly.shape)
print(poly.get_feature_names_out())
```

```python
# 데이터학습
lr=LinearRegression()
lr.fit(train_poly, train_label)

# 데이터 예측
predict= lr.predict(test_poly)
print('예측무게:',predict)

# 정확도
print('train 정확도:',lr.score(train_poly, train_label))
print('test 정확도:',lr.score(test_poly, test_label))
```

train정확도가 낮을 때 degree를 늘린다.
test정확도가 낮을 때 규제를 한다.

```python
# 규제하려면 정규화 후 규제 알고리즘 사용
# 정규화 작업을 하지 않으면 규제를 할 수 없음
ss= StandardScaler()
ss.fit(train_poly)
train_scaler=ss.transform(train_poly)
test_scaler=ss.transform(test_poly)
```

```python
# 규제 알고리즘
ridge =Ridge()
ridge.fit(train_scaler, train_label)

# 정확도
print('train 정확도:',ridge.score(train_scaler, train_label))
print('test 정확도:',ridge.score(test_scaler, test_label))
```
