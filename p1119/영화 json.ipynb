```python
import requests
import pandas as pd
```

```python
# 1. aip 주소
url="http://kobis.or.kr/kobisopenapi/webservice/rest/movie/searchMovieList.json?key=74d37f896fe094c9a03e563e8bb96a5e&itemPerPage=100"
```

```python
# 2. 데이터 불려오기
data = requests.get(url).json()
```

```python
# 3. 영화목록 추출 : data는 키 ["movieListResult"] 안에 키["movieList"]을 뽑는다
movies = data["movieListResult"]["movieList"]
```

```python
# 4. DataFrame 변환 (전체 컬럼)
df=pd.DataFrame(movies)
```

```python
# 4-1 감독이름 추출

df["directors"] = df["directors"].apply(
    lambda x: ", ".join(d["peopleNm"] for d in x)
)
```

```python
# 4-2 회사이름 추출
def companys(x):
    return ','.join(d["companyNm"] for d in x)

df['companys']=df["companys"].apply(companys)
```

```python
# 5. csv파일 저장 : encoding="utf-8-sig"-> 엑셀에서 csv파일을 열때 안깨지게 해주는것.

df.to_csv("C:/다운/데이터/movie_list.csv",index=False, encoding="utf-8-sig")
```
